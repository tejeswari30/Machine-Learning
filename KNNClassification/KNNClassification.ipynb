```python
import pandas as pd

# 1. Read the data
data = pd.read_csv(r"C:\Users\tejeswari\Desktop\DML\iris.csv")


# 2. Separate independent features (X) and target feature (y)
x = data.iloc[:, :-1].to_numpy()
y = data.iloc[:, -1].to_numpy()


# 3. Split the data into training and testing
from sklearn.model_selection import train_test_split

xtrain, xtest, ytrain, ytest = train_test_split(
    x, y, test_size=0.2, random_state=2
)


# 4. Model selection
from sklearn.neighbors import KNeighborsClassifier

model = KNeighborsClassifier(n_neighbors=2)


# 5. Train the model
model.fit(xtrain, ytrain)


# 6. Predict on test data
ypred = model.predict(xtest)


# 7. Calculate accuracy, precision and recall
from sklearn.metrics import accuracy_score, precision_score, recall_score

print('Accuracy: ', accuracy_score(ytest, ypred) * 100)
print('Precision: ', precision_score(ytest, ypred, average='macro') * 100)
print('Recall: ', recall_score(ytest, ypred, average='macro') * 100)


# 8. Predict for new data
print(model.predict([[1.2, 2.3, 4.2, 0.8]]))
```
