```python
import pandas as pd

# 1. Read the data
data = pd.read_csv(r"C:\Users\tejeswari\Desktop\DML\forestfires.csv")


# Label Encoding
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()

data['month'] = le.fit_transform(data['month'])
data['day'] = le.fit_transform(data['day'])


# 2. Separate independent features (X) and target feature (y)
x = data.iloc[:, :-1]
y = data.iloc[:, -1]


# 3. Split the data into training and testing
from sklearn.model_selection import train_test_split

xtrain, xtest, ytrain, ytest = train_test_split(
    x, y, test_size=0.3, random_state=52
)


# 4. Model selection
from sklearn.linear_model import LinearRegression

model = LinearRegression()


# 5. Train the model
model.fit(xtrain, ytrain)


# 6. Predict on test data
ypred = model.predict(xtest)


# 7. Calculate evaluation metric - RMSE
from sklearn.metrics import mean_squared_error

error = mean_squared_error(ytest, ypred)
print(error ** 0.5)  # Root Mean Squared Error


# 8. Predict for new data
```
